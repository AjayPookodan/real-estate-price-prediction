```python
from pathlib import Path
import pandas as pd

# --------------------------------------------------
# 1. Load the original dataset
# --------------------------------------------------

project_root = Path("..")
data_path = project_root / "data" / "raw" / "housing_raw.csv"

df = pd.read_csv(data_path, low_memory=False)

print("Dataset loaded successfully.")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])


# --------------------------------------------------
# 2. Describe the dataset columns
# --------------------------------------------------

column_info = {
    "PROP_ID": (
        "Unique property identifier",
        "identifier",
        "Not applicable",
        "Not normally used as a prediction feature"
    ),
    "PRICE": (
        "Listed property price",
        "text/numeric",
        "INR; verify units and transaction type",
        "Candidate target; raw values may contain Cr, lakh, or commas"
    ),
    "MIN_PRICE": (
        "Minimum listed price, if applicable",
        "text/numeric",
        "Verify from source",
        "Check meaning and consistency with PRICE"
    ),
    "MAX_PRICE": (
        "Maximum listed price, if applicable",
        "text/numeric",
        "Verify from source",
        "Check meaning and consistency with PRICE"
    ),
    "PRICE_PER_UNIT_AREA": (
        "Price per unit of property area",
        "text/numeric",
        "Verify from source",
        "Check units before using"
    ),
    "PRICE_SQFT": (
        "Price per square foot",
        "text/numeric",
        "INR/sqft, subject to verification",
        "Potentially derived from price and area; check for target leakage"
    ),
    "PROPERTY_TYPE": (
        "Type or category of property",
        "categorical",
        "Not applicable",
        "Potential prediction feature"
    ),
    "CITY": (
        "City associated with the property",
        "categorical",
        "Not applicable",
        "Verify geographic coverage from actual values"
    ),
    "LOCALITY": (
        "Locality or neighbourhood of the property",
        "categorical/text",
        "Not applicable",
        "May help represent location"
    ),
    "LOCALITY_WO_CITY": (
        "Locality name without the city name",
        "categorical/text",
        "Not applicable",
        "Check overlap with LOCALITY"
    ),
    "TRANSACT_TYPE": (
        "Transaction category",
        "categorical",
        "Not applicable",
        "Inspect values to distinguish sale, rent, and other listings"
    ),
    "PREFERENCE": (
        "Listing preference or category",
        "categorical",
        "Not applicable",
        "Meaning should be verified from source documentation"
    ),
    "OWNTYPE": (
        "Property ownership category",
        "categorical",
        "Not applicable",
        "Verify category definitions"
    ),
    "BEDROOM_NUM": (
        "Number of bedrooms",
        "numeric",
        "count",
        "Potential prediction feature"
    ),
    "BATHROOM_NUM": (
        "Number of bathrooms",
        "numeric",
        "count",
        "Potential prediction feature if present"
    ),
    "BALCONY_NUM": (
        "Number of balconies",
        "numeric",
        "count",
        "Potential prediction feature if present"
    ),
    "AREA": (
        "Listed property area",
        "text/numeric",
        "Verify from source",
        "Inspect for ranges and inconsistent formats"
    ),
    "CARPET_SQFT": (
        "Carpet area",
        "numeric/text",
        "square feet",
        "Check missing values and units"
    ),
    "BUILTUP_SQFT": (
        "Built-up area",
        "numeric/text",
        "square feet",
        "Check missing values and units"
    ),
    "SUPERBUILTUP_SQFT": (
        "Super built-up area",
        "numeric/text",
        "square feet",
        "Check missing values and units"
    ),
    "SUPER_SQFT": (
        "Super area measurement",
        "numeric/text",
        "square feet, subject to verification",
        "Check whether this duplicates another area column"
    ),
    "MIN_AREA_SQFT": (
        "Minimum listed area",
        "numeric/text",
        "square feet",
        "May be part of an area range"
    ),
    "MAX_AREA_SQFT": (
        "Maximum listed area",
        "numeric/text",
        "square feet",
        "May be part of an area range"
    ),
    "FURNISH": (
        "Furnishing category",
        "categorical",
        "Not applicable",
        "Verify category values"
    ),
    "FACING": (
        "Direction the property faces",
        "categorical",
        "Not applicable",
        "Potential prediction feature"
    ),
    "AGE": (
        "Property age or age category",
        "numeric/categorical",
        "Years or category; verify from source",
        "Inspect format before converting"
    ),
    "FLOOR_NUM": (
        "Floor on which the property is located",
        "numeric/text",
        "floor number",
        "Inspect non-numeric values"
    ),
    "TOTAL_FLOOR": (
        "Total number of floors in the building",
        "numeric/text",
        "count",
        "Potential prediction feature"
    ),
    "FEATURES": (
        "Property features or facilities",
        "text",
        "Not applicable",
        "May contain multiple features in one field"
    ),
    "AMENITIES": (
        "Amenities associated with the property",
        "text",
        "Not applicable",
        "May be useful for feature engineering"
    ),
    "TOP_USPS": (
        "Highlighted property selling points",
        "text",
        "Not applicable",
        "May contain multiple values"
    ),
    "TOTAL_LANDMARK_COUNT": (
        "Number of landmarks recorded for the property",
        "numeric",
        "count",
        "Verify how landmarks were counted"
    ),
    "FORMATTED_LANDMARK_DETAILS": (
        "Formatted information about nearby landmarks",
        "text",
        "Not applicable",
        "Inspect whether distances or landmark categories are available"
    ),
    "MAP_DETAILS": (
        "Map-related property information",
        "text/structured",
        "Not applicable",
        "Inspect before extracting location features"
    ),
    "location": (
        "Location-related information",
        "text/structured",
        "Not applicable",
        "Inspect format; may contain useful geographic information"
    ),
    "CITY_ID": (
        "Identifier associated with the city",
        "identifier",
        "Not applicable",
        "May duplicate information in CITY"
    ),
    "BUILDING_ID": (
        "Identifier associated with the building",
        "identifier",
        "Not applicable",
        "Not normally a direct prediction feature"
    ),
    "SOCIETY_NAME": (
        "Name of the residential society",
        "text",
        "Not applicable",
        "May contain missing values or high-cardinality categories"
    ),
    "BUILDING_NAME": (
        "Name of the building",
        "text",
        "Not applicable",
        "May contain high-cardinality values"
    ),
    "DESCRIPTION": (
        "Textual description of the property",
        "text",
        "Not applicable",
        "Unstructured text"
    ),
    "PROP_NAME": (
        "Property name",
        "text",
        "Not applicable",
        "Check overlap with other name fields"
    ),
    "PROP_HEADING": (
        "Property listing heading",
        "text",
        "Unstructured text"
    ),
    "PROP_DETAILS_URL": (
        "URL for property details",
        "text/URL",
        "Not applicable",
        "Source listing reference"
    ),
    "PHOTO_URL": (
        "URL for the property's photo",
        "text/URL",
        "Not applicable",
        "Image reference, not a numeric feature"
    ),
    "MEDIUM_PHOTO_URL": (
        "URL for a medium-sized property photo",
        "text/URL",
        "Not applicable",
        "Image reference"
    ),
    "PROPERTY_IMAGES": (
        "Property image information",
        "text/structured",
        "Not applicable",
        "May contain multiple image references"
    ),
    "THUMBNAIL_IMAGES": (
        "Thumbnail image information",
        "text/structured",
        "Not applicable",
        "May contain multiple image references"
    ),
    "REGISTER_DATE": (
        "Property registration date recorded in the dataset",
        "date/text",
        "Date",
        "Verify date format and meaning"
    ),
    "POSTING_DATE": (
        "Date the listing was posted",
        "date/text",
        "Date",
        "May help identify the dataset's time coverage"
    ),
    "UPDATE_DATE": (
        "Date the listing was updated",
        "date/text",
        "Date",
        "May help identify the dataset's time coverage"
    ),
    "EXPIRY_DATE": (
        "Listing expiry date",
        "date/text",
        "Date",
        "Verify meaning and availability"
    ),
    "FURNISHING_ATTRIBUTES": (
        "Detailed furnishing attributes",
        "text/structured",
        "Not applicable",
        "May contain mixed data types"
    ),
    "COMMON_FURNISHING_ATTRIBUTES": (
        "Common furnishing attributes",
        "text/structured",
        "Not applicable",
        "May contain multiple values"
    ),
    "QUALITY_SCORE": (
        "Quality score associated with the listing",
        "numeric",
        "Score; verify scale",
        "Check how the score was calculated"
    ),
    "BROKERAGE": (
        "Brokerage-related listing information",
        "text/numeric",
        "Verify from source",
        "Check units and meaning"
    ),
    "CONTACT_NAME": (
        "Name of the listing contact",
        "text",
        "Not applicable",
        "Personal/contact information; exclude from prediction features"
    ),
    "CONTACT_COMPANY_NAME": (
        "Company name associated with the contact",
        "text",
        "Not applicable",
        "Check relevance before using"
    ),
    "DEALER_PHOTO_URL": (
        "URL for the dealer's photo",
        "text/URL",
        "Not applicable",
        "Not normally a prediction feature"
    ),
    "metadata": (
        "Additional metadata",
        "text/structured",
        "Not applicable",
        "Inspect contents before using"
    ),
    "profile": (
        "Profile-related listing information",
        "text/structured",
        "Not applicable",
        "Meaning should be verified"
    ),
    "xid": (
        "External or source-specific identifier",
        "identifier",
        "Not applicable",
        "Not normally a direct prediction feature"
    ),
    "CLASS": (
        "Source-specific property classification",
        "categorical/text",
        "Not applicable",
        "Verify meaning from source documentation"
    ),
    "CLASS_HEADING": (
        "Heading associated with the property classification",
        "text",
        "Not applicable",
        "Verify meaning from source documentation"
    ),
    "CLASS_LABEL": (
        "Label associated with the property classification",
        "categorical/text",
        "Not applicable",
        "Verify category definitions"
    ),
    "SECONDARY_TAGS": (
        "Additional tags associated with the listing",
        "text/structured",
        "Not applicable",
        "May contain multiple categories"
    ),
    "GROUP_NAME": (
        "Group name associated with the listing",
        "text",
        "Not applicable",
        "Verify meaning from source documentation"
    ),
    "LISTING": (
        "Listing-related information or category",
        "text/categorical",
        "Not applicable",
        "Verify meaning from source documentation"
    ),
    "RES_COM": (
        "Residential/commercial classification or indicator",
        "categorical/text",
        "Not applicable",
        "Verify actual values and meaning"
    ),
    "PROPERTY_NUMBER": (
        "Property number or source-specific identifier",
        "text/identifier",
        "Not applicable",
        "Not normally a direct prediction feature"
    ),
    "MIN_PRICE": (
        "Minimum price associated with a listing",
        "text/numeric",
        "Verify from source",
        "Check for overlap with PRICE"
    ),
    "MAX_PRICE": (
        "Maximum price associated with a listing",
        "text/numeric",
        "Verify from source",
        "Check for overlap with PRICE"
    ),
}



# --------------------------------------------------
# 3. Generate the data dictionary
# --------------------------------------------------

dictionary_rows = []

for col in df.columns:

    # Default information for columns not documented yet
    default_info = (
        "Meaning not yet verified",
        str(df[col].dtype),
        "Not verified",
        "Inspect sample values and source documentation"
    )

    info = column_info.get(col, default_info)

    # Fix entries that do not contain exactly four values
    if not isinstance(info, (tuple, list)) or len(info) != 4:
        print(
            f"Warning: Invalid dictionary entry for '{col}'. "
            f"Using default information."
        )
        info = default_info

    meaning, data_type, unit, notes = info

    # Calculate missing percentage
    missing_pct = df[col].isna().mean() * 100

    # Record actual pandas data type
    actual_dtype = str(df[col].dtype)
    notes = f"{notes}; pandas dtype: {actual_dtype}"

    # Escape special characters for Markdown
    def escape_markdown(value):
        return (
            str(value)
            .replace("|", "\\|")
            .replace("\n", " ")
        )

    dictionary_rows.append({
        "Column": escape_markdown(col),
        "Meaning": escape_markdown(meaning),
        "Type": escape_markdown(data_type),
        "Unit": escape_markdown(unit),
        "Missing %": f"{missing_pct:.2f}%",
        "Notes": escape_markdown(notes)
    })


# Convert to DataFrame
dictionary = pd.DataFrame(dictionary_rows)


# --------------------------------------------------
# 4. Save the Markdown file
# --------------------------------------------------

project_root = Path("..")

reports_dir = project_root / "reports"
reports_dir.mkdir(parents=True, exist_ok=True)

dictionary_path = reports_dir / "data_dictionary.md"


# Create Markdown content
markdown = f"""# Data Dictionary

## Dataset Overview

- **Dataset file:** `data/raw/housing_raw.csv`
- **Number of rows:** {df.shape[0]:,}
- **Number of columns:** {df.shape[1]}
- **Purpose:** Document the fields in the raw real-estate dataset.
- **Note:** Column meanings and units marked for verification must be
  confirmed against the original dataset documentation.

## Column Descriptions

{dictionary.to_markdown(index=False)}

## Data Quality Notes

- The original dataset is preserved in `data/raw/`.
- Missing percentages are calculated from the raw data before cleaning.
- Data types represent the current pandas representation of each column.
- Price units and transaction categories must be verified before
  preparing the prediction target.
- Listed prices should not automatically be interpreted as completed
  sale prices.
"""

# Write file
dictionary_path.write_text(markdown, encoding="utf-8")


# --------------------------------------------------
# 5. Confirm successful creation
# --------------------------------------------------

print("\nData dictionary created successfully!")
print("File location:", dictionary_path.resolve())
print("Columns documented:", len(dictionary))
print("File size:", dictionary_path.stat().st_size, "bytes")

print("\nFirst five entries:")
display(dictionary.head())

```

```python
from pathlib import Path
from geopy.geocoders import Nominatim

# Project directories
project_root = Path("..")
raw_dir = project_root / "data" / "raw"
raw_dir.mkdir(parents=True, exist_ok=True)

cache_path = raw_dir / "geocode_cache.json"

# Initialize geocoder
geolocator = Nominatim(
    user_agent="real_estate_price_prediction_college_project"
)

# Load previously cached coordinates
if cache_path.exists():
    with open(cache_path, "r", encoding="utf-8") as f:
        cache = json.load(f)
else:
    cache = {}

print("Cached locations:", len(cache))
```

```python
import json
import time

CITY = "Gurgaon, Haryana, India"

def geocode_locality(name):
    name = str(name).strip()

    if not name:
        return None

    # Reuse previously obtained coordinates
    if name in cache:
        return cache[name]

    try:
        result = geolocator.geocode(
            f"{name}, {CITY}",
            timeout=15
        )

        if result:
            coordinates = [
                result.latitude,
                result.longitude
            ]

            cache[name] = coordinates

        else:
            # Cache a not-found result to avoid repeating it
            cache[name] = None
            coordinates = None

        # Save after each completed request
        with open(cache_path, "w", encoding="utf-8") as f:
            json.dump(cache, f, indent=2)

        # Respect the public service's request rate
        time.sleep(1.1)

        return coordinates

    except Exception as e:
        # Do not cache errors; they can be retried later
        print(f"Error geocoding '{name}': {e}")
        return None
```

```python
print("Starting geocoding test...")

localities = (
    df[locality_col]
    .dropna()
    .astype(str)
    .str.strip()
)

localities = localities[localities.ne("")].unique()

print("Unique localities:", len(localities))
print("Already cached:", len(cache))

for locality in localities[:5]:
    print(f"\nLooking up: {locality}", flush=True)

    coordinates = geocode_locality(locality)

    print(f"Result: {coordinates}", flush=True)

print("\nGeocoding test finished.")
```

```python
# Geocode all unique localities

total = len(localities)

for i, locality in enumerate(localities, start=1):

    if locality in cache:
        continue

    print(f"[{i}/{total}] Geocoding: {locality}", flush=True)

    coordinates = geocode_locality(locality)

    print(f"    Coordinates: {coordinates}", flush=True)

print("\nGeocoding completed!")
print("Total localities:", total)
print("Cached results:", len(cache))
print("Successful results:", sum(v is not None for v in cache.values()))
print("Unsuccessful lookups:", sum(v is None for v in cache.values()))
```

```python
# Convert successful cached coordinates into a DataFrame

coords = pd.DataFrame(
    [
        (locality, values[0], values[1])
        for locality, values in cache.items()
        if values is not None
    ],
    columns=["LOCALITY", "latitude", "longitude"]
)

# Ensure one row per locality
coords = coords.drop_duplicates(subset=["LOCALITY"])

# Remove previously merged coordinate columns if present
df = df.drop(
    columns=["latitude", "longitude"],
    errors="ignore"
)

# Merge coordinates with the original property records
df = df.merge(
    coords,
    on="LOCALITY",
    how="left",
    validate="many_to_one"
)

print("Total property records:", len(df))
print("Unique localities:", df["LOCALITY"].nunique())
print("Localities with coordinates:", coords["LOCALITY"].nunique())
print("Properties without coordinates:", df["latitude"].isna().sum())

display(df[["LOCALITY", "latitude", "longitude"]].head(10))
```

```python
import folium
from pathlib import Path

# Keep rows with valid coordinate values
map_df = df.dropna(subset=["latitude", "longitude"]).copy()

# Use one marker per locality
map_df = map_df.drop_duplicates(subset=["LOCALITY"])

# Create the map
if not map_df.empty:

    center_lat = map_df["latitude"].mean()
    center_lon = map_df["longitude"].mean()

    m = folium.Map(
        location=[center_lat, center_lon],
        zoom_start=11
    )

    # Draw the approximate Gurgaon bounding box
    folium.Rectangle(
        bounds=[
            [LAT_MIN, LON_MIN],
            [LAT_MAX, LON_MAX]
        ],
        color="red",
        fill=False,
        tooltip="Approximate Gurgaon screening bounds"
    ).add_to(m)

    # Add a marker for each locality
    for _, row in map_df.iterrows():

        is_outside = (
            row["latitude"] < LAT_MIN
            or row["latitude"] > LAT_MAX
            or row["longitude"] < LON_MIN
            or row["longitude"] > LON_MAX
        )

        folium.CircleMarker(
            location=[
                row["latitude"],
                row["longitude"]
            ],
            radius=4,
            popup=(
                f"Locality: {row['LOCALITY']}<br>"
                f"Latitude: {row['latitude']}<br>"
                f"Longitude: {row['longitude']}<br>"
                f"Outside bounds: {is_outside}"
            ),
            color="red" if is_outside else "blue",
            fill=True
        ).add_to(m)

    # Save the HTML file
    project_root = Path("..")
    reports_dir = project_root / "reports"
    reports_dir.mkdir(parents=True, exist_ok=True)

    map_path = reports_dir / "locality_map.html"
    m.save(str(map_path))

    print("Interactive map saved successfully!")
    print("Map location:", map_path.resolve())
    print("Localities displayed:", len(map_df))

else:
    print("No valid coordinates available to plot.")
```
